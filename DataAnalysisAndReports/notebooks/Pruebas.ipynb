```python
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
import pandas as pd

df_Random_Forest2 = pd.read_csv('C:/Users/Adri/Desktop/Triage_HF/DataAnalysisAndReports/dataset/Pacientes2.csv')



# Separar las características (X) de la etiqueta objetivo (y)
X_randomforest = df_Random_Forest2.iloc[:,2:]
y_randomforest = df_Random_Forest2.iloc[:,1]
```

```python

# Dividir los datos en conjuntos de entrenamiento y prueba
X_train, X_test, y_train, y_test = train_test_split(X_randomforest, y_randomforest, test_size=0.8, random_state=0)

# Crear el modelo de árbol de decisión
clf = DecisionTreeClassifier()

# Entrenar el modelo
clf.fit(X_train, y_train)

# Hacer predicciones en el conjunto de prueba
y_pred = clf.predict(X_test)

# Calcular la precisión del modelo
accuracy = accuracy_score(y_test, y_pred)

print(f'La precisión del modelo de árbol de decisión es: {accuracy}')
```
